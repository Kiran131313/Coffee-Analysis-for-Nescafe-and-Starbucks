```python
#Write a Python program to create a list of 6 integers, print the second and
#fourth elements, and print a slice containing the third, fourth, and fifth elements.

numbers = [10, 20, 30, 40, 50, 60]
print("Second element:", numbers[1])
print("Fourth element:", numbers[3])
slice_3_to_5 = numbers[2:5]
print("Slice (3rd–5th elements):", slice_3_to_5)
```

```python
#Write a Python program to create a tuple with your name, city, and age,
#unpack it into three variables, and print each variable.

person = ("Kiran", "Bengaluru", 29)
name, city, age = person
print("Name:", name)
print("City:", city)
print("Age:", age)
```

```python
#3:Create a set with numbers (including duplicates), print the set, and check if 10
# is present in the set.

numbers = {1, 5, 7, 10, 5, 3, 10}
print("Set contents:", numbers)
if 10 in numbers:
    print("10 is in the set")
else:
    print("10 is not in the set")
```

```python
#4: Create a tuple with repeated elements, convert it to a set, and print both the
#tuple and the set.

my_tuple = (1, 2, 2, 3, 4, 4, 5)
print("Original tuple:", my_tuple)
my_set = set(my_tuple)
print("Converted set:", my_set)
```
